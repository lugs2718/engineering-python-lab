```python
from transformers import BertTokenizer, BertModel

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')
```

```python
inputs = tokenizer("rainha", return_tensors="pt")
inputs0 = tokenizer("mulher", return_tensors="pt")
inputs1 = tokenizer("rei", return_tensors="pt")
inputs2 = tokenizer("homem", return_tensors="pt")
outputs = model(**inputs)
outputs0 = model(**inputs0)
outputs1 = model(**inputs1)
outputs2 = model(**inputs2)

# Vetor da frase (embedding do token [CLS])
sentence_vector = outputs.last_hidden_state[:, 0, :]
sentence_vector0 = outputs0.last_hidden_state[:, 0, :]
sentence_vector1 = outputs1.last_hidden_state[:, 0, :]
sentence_vector2 = outputs2.last_hidden_state[:, 0, :]
```
