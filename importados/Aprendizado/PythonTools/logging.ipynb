```python
import logging

# Configure the logging module
logging.basicConfig(
    level=logging.DEBUG, # Set the minimum level of messages to capture
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    handlers=[
        logging.StreamHandler(), # Log messages to the console
        logging.FileHandler('app.log') # Log messages to a file named 'app.log'
    ]
)

# Get a logger instance
logger = logging.getLogger(__name__)

# Test different logging levels
logger.debug('This is a debug message.')
logger.info('This is an info message.')
logger.warning('This is a warning message.')
logger.error('This is an error message.')
logger.critical('This is a critical message.')

print("Check your console for log messages and a new file named 'app.log' in your Colab environment for the log output.")
```

This code sets up basic logging:
- It configures `logging` to capture messages from `DEBUG` level and above.
- It formats the log messages to include timestamp, logger name, level, and the message itself.
- It directs log output to both the console (StreamHandler) and a file named `app.log` (FileHandler).
- It then demonstrates messages at different logging levels (DEBUG, INFO, WARNING, ERROR, CRITICAL).
