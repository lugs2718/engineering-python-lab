```python
import os
import shutil

origem = "/content/drive/MyDrive/1.aulas-exercicios-aprendizado"
destino = "/content/engineering-python-lab/importados"
extensoes_validas = (".py", ".ipynb", ".md", ".txt", ".sql", ".json", ".csv", ".xlsx")

for raiz, pastas, arquivos in os.walk(origem, followlinks=True):
  # Ignora pastas ocultas como .git ou .ipynb_checkpoints
  pastas[:] = [p for p in pastas if not p.startswith(".")]

  for arquivo in arquivos:
    if arquivo.endswith(extensoes_validas):
      caminho_completo = os.path.join(raiz, arquivo)
      # Mantém a estrutura relativa de pastas
      rel_path = os.path.relpath(raiz, origem)
      pasta_destino = os.path.join(destino, rel_path)
      os.makedirs(pasta_destino, exist_ok=True)

      shutil.copy2(caminho_completo, os.path.join(pasta_destino, arquivo))
      print(f"Copiado: {arquivo}")
```
