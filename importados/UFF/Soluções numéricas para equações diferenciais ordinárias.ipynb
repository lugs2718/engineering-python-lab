#Método de Euler

##Consiste em aproximar a função solução da equação diferencial por meio de uma  reta cuja inclinação é igual à desta função.

##$\dot{y} = f(t,y)$

#Método de Heun

##Este método, apesar de utilizar uma aproximação similar ao método de Euler, antes faz uma média aritmética das incliniações nos extremos do intervalo [$x_i, x_{i+1}$] sendo $x_{i+1} = x_i + h$. Isto é,

##<center>$I_m = \frac{h}{2}(f(x_i,y_i) + f(x_{i+1},y_{i+1}))$<center/>

##Para a primeira iteração, utiliza-se o método de Euler para aproximar $y_{i+1}$

```python
#EDO

def func(t,y):
  return -y

#PVI
t_0 = 0
y_0 = 1

#Interalo
a = float(input('insira o limite inferior : '))
b = float(input('insira o limite superior : '))

if a>b:
  raise Exception('O limite superior deve ser maior queo limite inferior')

I = [a, b]

#Número de subintervalos
n = int(input('Insira o número de subintervalos : '))

#Método de Heun
t = [t_0]
y = [y_0]

e = []
e.append(exp(-t_0))

err = []# erro de aproximação

h = (I[1] - I[0])/n

for i in range(n):
  k1 = func(t_0, y_0)
  k2 = func(t_0 + h, y_0 + h*k1)

  y_i = y_0 + h*0.5*(k1+k2)
  y.append(y_i)

  t_0 += h
  t.append(t_0)

  y_0 = y_i


for i in range(n):
  e.append(exp(-t[i+1]))
  err.append(abs(e[i]-y[i]))

  print(f'y({t[i]}) = {y[i]}')
  print(f'exp({t[i]}) = {e[i]}')
  print(f'o erro cometido foi de {err[i]}')
  print('-'*10)

plt.plot(t,y)
plt.plot(t,e)
plt.show()
```
