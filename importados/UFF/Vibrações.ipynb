```python
#RODE ESTA CÉLULA PRIMEIRO
import math #biblioteca que permite acessar o valor de pi no python
```

```python
#Dados
Keq=4e+4
c=2e+3
M=1585
v=110/3.6
lambd=6
Y=0.01


#frequências(rad/s)
W=(2*math.pi*v)/lambd
Wn=(Keq/M)**(1/2)

r=W/Wn
Cc=2*((Keq*M)**(1/2))
csi=c/Cc

num=(1+((2*csi*r)**2))**(1/2)
den=((((1-(r**2))**2)+((2*csi*r)**2))**(1/2))
razao=num/den

result = {
    'w':W,
    'wn':Wn,
    'r': r,
    'Cc':Cc,
    'csi':csi,
    'num':num,
    'den':den,
    'razao': razao
}
print(result)
```
