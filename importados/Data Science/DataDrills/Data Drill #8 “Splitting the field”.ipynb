# Task

Your task is to aggregate this data and return a table counting the number of players at each position, sorted from highest to lowest.

# Minha solução

```python
import pandas as pd
data = pd.read_csv(r'..\datasets\baseball_positions.csv')
data
```

```python
pd.set_option('display.max_rows', None)
```

```python
# 1. Create a columns which contains lists of all positions of each player
baseball_df = data.copy()
player_positions = baseball_df.Position.str.split('/')

baseball_df['Position_split'] = player_positions

# 2. Explode the list from the position_split column, so each position is stored in different rows
baseball_df_exploded = baseball_df.explode(column='Position_split')
baseball_df_exploded

# 3. Agregate and count the number of players in each position
player_positions_grouped = baseball_df_exploded.groupby('Position_split').count().sort_values(by='Position',ascending=False)
player_positions_grouped
```
