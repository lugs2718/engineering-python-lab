```python
import pandas as pd
coffee_df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/datasets/coffee_shop_sales.csv', parse_dates=['date'])
coffee_df.head()
```

```python
coffee_df_analysis = coffee_df.copy()

# Create month column
coffee_df_analysis['month'] = coffee_df_analysis['date'].dt.month

coffee_df_analysis.pivot_table(index=['store', 'month'], values = 'sales', aggfunc='sum').assign(
    over_to_month = lambda x: x.diff(periods=1)
)
```
