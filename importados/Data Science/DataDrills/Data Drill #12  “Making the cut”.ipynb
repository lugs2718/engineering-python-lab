```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('..\datasets\marathon-data.csv')
df_analysis = data.copy()
df_analysis.final = pd.to_datetime(df_analysis.final)
df_analysis.head()
```

```python
df_analysis = data.copy()
df_analysis['category'] = df_analysis.final.apply(
    func=lambda x: 'Sub 3' if x<pd.Timestamp('03:00') else \
    '3:00 - 3:30' if x>=pd.Timestamp('03:00') and x<pd.Timestamp('03:30') else \
    '3:30 - 4:00' if x>=pd.Timestamp('03:30') and x<pd.Timestamp('04:00') else \
    '4:00 - 4:30' if x>=pd.Timestamp('04:00') and x<pd.Timestamp('04:30') else \
    '4:30 - 5:00' if x>=pd.Timestamp('04:30') and x<pd.Timestamp('05:00') else \
    '5:00 - 5:30' if x>=pd.Timestamp('05:00') and x<pd.Timestamp('05:30') else \
    '5:30 - 6:00' if x>=pd.Timestamp('05:30') and x<pd.Timestamp('06:00') else \
    '6:00+' if x>=pd.Timestamp('06:00') else None
)
```

```python
percentages = df_analysis.groupby('category').gender.count() / df_analysis.shape[0]
percentages*100
```

# Solução do vídeo

```python
marathon_df = data.copy()

# Converting the final times to total minutes
marathon_df['final_minutes'] = pd.to_timedelta(marathon_df['final']).dt.total_seconds() / 60

# Defining the categories

band_edges = [0, 180, 210, 240, 270, 300, 330, 360, float('inf')]
band_labels = ['sub 3', '3-3:30', '3:30-4', '4-:4:30', '4:30-5', '5-5:30', '5:30-6', '6+']

# Creating a band column to categorize the minutes
marathon_df['band'] = pd.cut(
    marathon_df['final_minutes'],
    bins=band_edges,
    labels=band_labels,
    right=False
)

result = marathon_df.band.value_counts(normalize=True) * 100
result
```
