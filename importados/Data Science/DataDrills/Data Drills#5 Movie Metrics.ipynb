You've been given a table of Netflix users and another with their viewing activity, including the movie name, date started, and whether they finished it.

Your task is to engineer these new features for each user, based on their activity:

    Date from the first movie they finished

    Name of the first movie they finished

    Date from the last movie they finished

    Name of the last movie they finished

    Movies started

    Movies finished

```python
import pandas as pd
```

```python
users_df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/datasets/users.csv')
activity_df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/datasets/activity.csv')
```

```python
users_df
```

```python
activity_df[(activity_df['user_id'] ==20) & (activity_df['finished'] == 1)].sort_values(by='date', ascending=False)
```

```python
last_and_first_date = activity_df.query('finished == 1').groupby('user_id').agg(name_first_movie_finished=('movie_name', 'first'), name_last_movie_finished=('movie_name', 'last'),date_last_movie_finished=('date', 'max'), date_first_movie_finished=('date', 'min'))
last_and_first_date
```

```python
movie_count = activity_df.groupby('user_id').agg(movies_started=('movie_name', 'count'), movies_finished=('finished', 'sum'))
movie_count
```

```python
final_df = pd.concat([last_and_first_date, movie_count], axis=1)
final_df
```

```python
# Return all rows with aggregate
activity_df.query('finished==1').sort_values(by='date', ascending=False).groupby('user_id').agg(lambda x: list(x)[0])
```
