## Import the relevant libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
import seaborn as sns
sns.set()
```

## Load the data

```python
data = pd.read_csv('real_estate_price_size_year_view.csv')
```

```python
data
```

## Create a dummy variable for 'view'

```python
data['view_dummy'] = data['view'].map({
    'Sea view': 1,
    'No sea view': 0
})
```

## Create the regression

### Declare the dependent and the independent variables

```python
X = data[['size', 'year', 'view_dummy']]
y = data['price']
```

### Regression

```python
sm.add_constant(X)
results = sm.OLS(y, X).fit()
results.summary()
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
from skleanr.preprocessing import Target

# 1. Dados de Exemplo (Formato Longo)
dados = pd.DataFrame({
    'ID_Plano': ['Plano_A', 'Plano_A', 'Plano_B', 'Plano_C'],
    'Idade_Maq': [5000, 5000, 2000, 10000],
    'ID_Material': ['OLEO_15W', 'FILTRO_AR', 'OLEO_15W', 'CORREIA_V'], # Coluna de Texto
    'Qtd_Consumida': [20, 1, 20, 2] # Target
})

# 2. Criando o Codificador
le = LabelEncoder()

# 3. Transformando o Texto em Números
# O modelo vai aprender uma coluna nova chamada "ID_Material_Encoded"
dados['ID_Material_Encoded'] = le.fit_transform(dados['ID_Material'])

# Veja como ficou a conversão
print("Mapeamento dos Materiais:")
for i, item in enumerate(le.classes_):
    print(f"{item} --> {i}")

print("\nDados prontos para o Random Forest:")
print(dados[['ID_Material', 'ID_Material_Encoded', 'Qtd_Consumida']])

# 4. Treinamento Simples
X = dados[['Idade_Maq', 'ID_Material_Encoded']] # Usamos o Encoded, não o texto
y = dados['Qtd_Consumida']

modelo = RandomForestRegressor()
modelo.fit(X, y)

print("\nModelo treinado com sucesso!")
```
