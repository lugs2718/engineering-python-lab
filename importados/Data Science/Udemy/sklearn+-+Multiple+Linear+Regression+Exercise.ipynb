## Import the relevant libraries

```python
import pandas as pd
import matplotlib as plt
import numpy as np
import seaborn as sb
sb.set()

from sklearn.linear_model import LinearRegression
```

## Load the data

```python
data = pd.read_csv('real_estate_price_size_year.csv')
```

```python
data
```

## Create the regression

### Declare the dependent and the independent variables

```python
X = data[['size', 'year']]
y = data['price']
```

### Regression

```python
reg = LinearRegression()
reg.fit(X, y)
```

### Find the intercept

```python
reg.intercept_
```

### Find the coefficients

```python
reg.coef_
```

### Calculate the R-squared

```python
r2 = reg.score(X, y)
r2
```

### Calculate the Adjusted R-squared

```python
n = X.shape[0]
p = X.shape[1]

adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
adj_r2
```

### Compare the R-squared and the Adjusted R-squared

The adjusted $R^2$ is just slightly smaller than $R^2$, so the addition of the new variable is not prejudicial to the predictive power of the model.

### Compare the Adjusted R-squared with the R-squared of the simple linear regression

By adding an extra variable that is also related to the response $y$, the predictive power of the model increased, and so has $R^2$.

```python
adj_X = X['size'].values.reshape(-1,1)
adj_y = y.values.reshape(-1,1)
```

```python
reg_slr = LinearRegression()
reg_slr.fit(adj_X, adj_y)
reg_slr.score(adj_X, adj_y)
```

### Making predictions

Find the predicted price of an apartment that has a size of 750 sq.ft. from 2009.

```python
reg.predict([[750, 2009]])
```

### Calculate the univariate p-values of the variables

```python
from sklearn.feature_selection import f_regression
```

```python
f_reg = f_regression(X, y)
```

```python
f_reg
```

### Create a summary table with your findings

```python
summary = pd.DataFrame({
  'Features' : X.columns,
  'Coefficients' : reg.coef_,
  'P-values' : f_reg[1].round(3)
})
display(summary)
```
