# Manipulating text files with python open() function

## Writing the text file

To write files, we use the writing mode ("w").

We must be careful not to lose the contents of the file, as the write mode **overwrite** the file.

```python
filename = 'test.txt'
texto = "Olá, meu nome é Luiz Gustavo \n Sou uma pessoa muito legal"

# Using open(), the file closes automatically
with open(filename, mode="w") as in_file:
  in_file.write(texto)
```

```python
print(in_file) # Retorna um wrapper, para retornar novamente o texto, usamos file.read()
```

## Reading the file

```python
with open(filename, mode="r") as out_file:
  file_content = out_file.read()
```
