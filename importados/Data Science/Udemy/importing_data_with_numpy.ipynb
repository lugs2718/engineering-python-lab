# Importing text data with Numpy

```python
import numpy as np
```

## np.loadtxt() vs np.genfromtxt()

- np.loadtxt(): "load" implies the data is ready to be imported and used
  - faster

- np.genfromtxt(): "generate" indicates that the function creates the dataset from the text file, which requires constructing the array
  - more flexible

## Importing the data

### Using np.loadtxt()

```python
data_file_path = "/content/drive/MyDrive/Cursos/DataScience/Udemy/Data/Lending-Company-Numeric-Data.csv"
leading_co_data_numeric_1 = np.loadtxt(data_file_path, delimiter=",")
leading_co_data_numeric_1
```

## Trying to import data with missing values

```python
data_nan_file_path = "/content/drive/MyDrive/Cursos/DataScience/Udemy/Data/Lending-Company-Numeric-Data-NAN.csv"
leading_co_data_numeric_NAN = np.loadtxt(data_nan_file_path, delimiter=";", dtype=str)
leading_co_data_numeric_NAN
```

np.loadtxt() cannot import txt data with missing values directly.

It requires treating the values as string.

```python
leading_co_data_numeric_NAN_gentxt = np.genfromtxt(data_nan_file_path, delimiter=";")
leading_co_data_numeric_NAN_gentxt
```

np.genfromtxt() on the other hand can import the data as float, without adjustments.

## Partial cleaning while importing

### Initial data

```python
leading_co_data_numeric_NAN_gentxt
```

### Cleaning the data

Skipping headers and footer is sometimes useful

```python
leading_co_data_numeric_NAN_gentxt = np.genfromtxt(data_nan_file_path,
                                                   delimiter=";",
                                                   skip_header=2,
                                                   )
leading_co_data_numeric_NAN_gentxt
```

```python
leading_co_data_numeric_NAN_gentxt = np.genfromtxt(data_nan_file_path,
                                                   delimiter=";",
                                                   skip_footer=2,
                                                   )
leading_co_data_numeric_NAN_gentxt
```

we can also skip columns

```python
leading_co_data_numeric_NAN_gentxt = np.genfromtxt(data_nan_file_path,
                                                   delimiter=";",
                                                   usecols=(5,0,1) # The ordering matters
                                                   )
leading_co_data_numeric_NAN_gentxt
```

```python
leading_co_data_numeric_NAN_gentxt = np.genfromtxt(data_nan_file_path,
                                                   delimiter=";",
                                                   usecols=(5,0,1),
                                                   skip_header=2,
                                                   skip_footer=2)
leading_co_data_numeric_NAN_gentxt
```

it is possible also to unpack the values

```python
leading_co_data_5, leading_co_data_0, leading_co_data_1  = np.genfromtxt(data_nan_file_path,
                                                   delimiter=";",
                                                   usecols=(5,0,1),
                                                   skip_header=2,
                                                   skip_footer=2, unpack=True)
print(leading_co_data_5)
print(leading_co_data_0)
print(leading_co_data_1)
```
