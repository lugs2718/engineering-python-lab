## Import the relevant libraries

```python
from sklearn.linear_model import LinearRegression
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

## Load the data

```python
data = pd.read_csv('real_estate_price_size.csv')
data.head()
```

```python
plt.scatter(data['size'], data['price'])
```

## Create the regression

### Declare the dependent and the independent variables

```python
X = data['size']
y = data['price']
```

### Explore the data

```python
X.values.shape
y.values.shape
```

### Transform the inputs into a matrix (2D object)

```python
reshaped_x = X.values.reshape(100,1)
reshaped_y = y.values.reshape(100,1)
```

### Regression itself

```python
# Instanciar a classe de regressão

reg = LinearRegression()
reg.fit(reshaped_x,reshaped_y)
```

### Calculate the R-squared

```python
reg.score(reshaped_x,reshaped_y)
```

### Find the intercept

```python
reg.intercept_
```

### Find the coefficients

```python
reg.coef_
```

### Making predictions

You find an apartment online with a size of 750 sq.ft.

All else equal what should be its price according to the model?

```python
reg.predict([[750]])
```

```python
plt.scatter(data['size'], data['price'])
plt.xlabel('size', fontsize=20)
plt.ylabel('price', fontsize=20)
plt.plot(reshaped_x, reg.predict(reshaped_x), lw=4, c='orange', label='regression')
plt.plot
```
