```python
import pandas as pd

def second_highest_salary(employee: pd.DataFrame) -> pd.DataFrame:
    # Remove salary duplicates
    employee_no_duplicates=employee.drop_duplicates(subset=['salary'])

    # Sort table by salary
    employee_sorted=employee_no_duplicates.sort_values(by=['salary'],ascending=False)
    print(employee_sorted)

    # get the second highest
    try:
        second_highest=pd.DataFrame({'SecondHighestSalary':[employee_sorted.salary.iloc[1]]})
    except Exception:
        second_highest=pd.DataFrame({'SecondHighestSalary':[None]})

    return second_highest
```

```python
data = [[1, 100], [2, 200], [3, 300]]
employee = pd.DataFrame(data, columns=['id', 'salary']).astype({'id':'int64', 'salary':'int64'})
```

```python
second_highest_salary(employee)
```
